# E-Commerce Sales & Profit Analysis (Portfolio Project)
**Tools:** Python, Pandas, Matplotlib, Seaborn

This notebook performs data cleaning, exploratory data analysis (EDA), and visualization on the cleaned e-commerce dataset (`ecommerce_cleaned_data.csv`).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv('ecommerce_cleaned_data.csv')
df.head()
```

```python
# Check summary statistics and missing values
print(df.info())
print('\nMissing Values:\n', df.isnull().sum())
df.describe()
```

```python
# Category and Region Analysis
cat_summary = df.groupby('Category')[['Sales', 'Profit', 'Quantity']].sum().reset_index()
print(cat_summary)

plt.figure(figsize=(8, 5))
sns.barplot(data=cat_summary, x='Category', y='Sales', palette='viridis')
plt.title('Total Sales by Category')
plt.show()
```
